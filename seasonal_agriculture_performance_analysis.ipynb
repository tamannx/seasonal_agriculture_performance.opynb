```python
for col in ["Rainfall_mm","Soil_Moisture_pct","Yield_Tonnes_Ha"]:          #For each of these 3 columns, group the data by season, and fill any blank value with that season's own middle value (median)
    df[col]=df.groupby("Season")[col].transform(lambda x: x.fillna(x.median()))   #Why median, not average (mean)? Median is safer when a column has extreme values (very high or very low numbers) because it doesn't get pulled around by them the way an average does.
                                                                               #.transform() always returns a result with the same shape and index as the original data — one output value per input row. .apply() can return anything — a single value per group, a differently-shaped DataFrame, or a Series
```

# Analysis

```python
df.groupby("Season")[["Total_Cost_INR","Revenue_INR","Profit_INR"]].agg(["mean","median","std"]).round(2)   #Profit & Revenue by Season
                                                                                                            # a high std means huge variability
                                                                                                            # agg() with a list applies the same stats to every column listed — if you want different stats per column, you need a dict instead: 
                                                                                                            # .agg({'Profit_INR': ['mean', 'std'], 'Revenue_INR': ['mean']})
```

"Zaid season is the least profitable — not because costs are unusually high in isolation, but because farmers pay similar costs to other seasons while earning significantly less revenue. Revenue, not cost, is the primary driver of seasonal profit differences."

```python
df["loss"]=df["Profit_INR"]<0
df.groupby("Season")["loss"].mean()*100
```

Nearly 2 in 3 farms operate at a loss during Zaid season, compared to fewer than half in Kharif and half in Rabi.

```python
loss=df.groupby("Season")["loss"].mean()*100
plt.figure(figsize=(7,3))
ax=loss.plot(kind='barh')
plt.title('% of farms operating on loss by season')
plt.xlabel('% of farms operating on negative profit')
plt.bar_label(ax.containers[0],fmt="%.1f%%")
plt.tight_layout()
# plt.savefig()
```

```python
plt.figure(figsize=(6,5))
sns.boxplot(x="Season",y="Profit_INR",hue="Season",order=["Kharif","Rabi","Zaid"],data=df,legend=False,flierprops=dict(markerfacecolor="brown",markeredgecolor="black"))
plt.axhline(0,color="black",linestyle="--",linewidth=1)
plt.title("Profit distribution by sales")
plt.ylabel("Profit INR")
plt.tight_layout()
plt.show()
```

```python
df.groupby("Season")[["Yield_Tonnes_Ha","Production_Tonnes","Market_Price_INR_Tonne"]].agg(["mean","median","std"]).round(2).T
```

Market price remains stable across seasons (within 1.5%), ruling out price as a cause of seasonal profit differences. 
Instead, declining yield and production — down ~16-18% from Kharif to Zaid — are the direct drivers of Zaid's financial underperformance

```python
df.groupby("Season")[["Rainfall_mm","Avg_Temperature_C","Humidity_pct","Soil_Moisture_pct","Water_Used_m3","Water_Efficiency_t_per_1000m3"]].agg(["mean","median","std"]).round(2).T
```

Zaid season's poor profitability traces back to harsh growing conditions — 65% less rainfall and the highest average temperature of any season. 
Farmers respond by using more irrigation water than any other season, yet achieve the lowest water-use efficiency, resulting in lower yield, 
'lower production, and ultimately the highest farmer loss rate (65%) despite comparable farming costs.
